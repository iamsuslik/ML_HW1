7. Рассмотрим случайный вектор с формой 100х2, представляющей координаты, и найдем расстояния от точки к точке

````python
points = np.random.random((100, 2))

X = np.atleast_2d(points[:, 0])
Y = np.atleast_2d(points[:, 1])

dx = X - X.T
dy = Y - Y.T

dist_matrix = np.sqrt(dx**2 + dy**2)

print("Размер матрицы расстояний:", dist_matrix.shape)
print("Расстояние от 0-й точки до самой себя (всегда 0):", dist_matrix[0, 0])
````

Найдите ответы на вопросы или выполните действия с предложенным датасетом 

Файл: `russian-housing-market.csv`

````python
import pandas as pd
````

````python
# Load dataset here
df = pd.read_csv("russian-housing-market.csv")
````

1. Посчитайте количество записей в датасете с отсутствующей информацией о жилой площади жилого помещения

````python
missing_life_sq = df["life_sq"].isna().sum()
print(f"1. Пропусков в life_sq: {missing_life_sq}")
````

2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
# отсутствующие занчения заменяем медианой, т.к. она устойчива к выбросам.
# Медиана дает значение, которое ближе всего к площади среднестатистической квартиры,
# не искажая статистику экстремальными значениями
median_life = df["life_sq"].median()
df["life_sq"] = df["life_sq"].fillna(median_life)
````

3. Какие значения имеет 25, 50 и 75 перцентили полной площади помещения

````python
percentiles = df["full_sq"].quantile([0.25, 0.5, 0.75])
print(percentiles)
````

4. Посчитайте среднюю стоимость продажи помещений с общей площадью от 50 до 70 квадратных метров

````python
avg_price = df.query("full_sq >= 50 and full_sq <= 70")["price_doc"].mean()
print(f"Средняя стоимость: {avg_price}")
````

5. Какая максимальная этажность здания представлена в датасете

````python
max_floor = df["max_floor"].max()
print(f"Максимальная этажность: {max_floor}")
````

6. Выявите район с самым дорогим жильем

````python
expensive_district = df.groupby("sub_area")["price_doc"].median().idxmax()
print(f"Район с самым дорогим жильем: {expensive_district}")
````

7. Посчитайте среднюю стоимость за квадратный метр жилья в Митино(Mitino)

````python
mitino = df[df["sub_area"] == "Mitino"]
price_per_sq = (mitino["price_doc"] / mitino["full_sq"]).mean()
print(f"Средняя стоимость за квадратный метр жилья в Митино: {price_per_sq}")
````

8. Посчитайте среднюю площадь комнаты в Бибирево

````python
bibirevo = df[df["sub_area"] == "Bibirevo"].copy()

bibirevo = bibirevo[bibirevo["num_room"] > 0]

bibirevo = bibirevo.dropna(subset=["life_sq", "num_room"])

bibirevo["room_size"] = bibirevo["life_sq"] / bibirevo["num_room"]

avg_room_sq = bibirevo["room_size"].mean()

print(f"Средняя площадь комнаты в Бибирево: {avg_room_sq:.2f} кв. м.")
````

9. Проанализируйте есть ли разница в цене между квартирами на 1 и 9 этажах

````python
price_1 = df[df["floor"] == 1]["price_doc"].mean()
price_9 = df[df["floor"] == 9]["price_doc"].mean()
print(f"Разница: {price_9 - price_1}")
````

10. Найдите минимальную стоимость квартиры с жилой площадью не менее 60 квадратных метров, не менее чем 3мя комнатами и не ниже чем на 5 этаже

````python
filtered = df[(df["full_sq"] >= 60) & (df["num_room"] >= 3) & (df["floor"] >= 5)]
min_price = filtered["price_doc"].min()
print(f"Минимальная цена на квартиру с площадью не менее 60 м², 3+ комнат и на 5+ этаже: {min_price}")
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import matplotlib.pyplot as plt
import seaborn as sns

df['timestamp'] = pd.to_datetime(df['timestamp'])

print("Формат даты успешно изменен.")
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
df_nz = df[df["sub_area"] == "Nagatinskij Zaton"]
# df_nz.groupby("timestamp")["price_doc"].mean().plot()
df_nz.groupby(df_nz["timestamp"].dt.to_period("M"))["price_doc"].mean().plot(figsize=(10, 5))
plt.title("Динамика цен в Нагатинском Затоне")
plt.show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
mask = (df["full_sq"] < 50) & (df["timestamp"].dt.year.between(2011, 2014))
df[mask].groupby(df["timestamp"].dt.to_period("M"))["price_doc"].median().plot(figsize=(12, 6), marker='o')

plt.title("Изменение медианной цены на квартиры < 50 кв.м. (2011-2014)")
plt.xlabel("Дата")
plt.ylabel("Медианная цена")
plt.grid(True)
plt.show()
````
